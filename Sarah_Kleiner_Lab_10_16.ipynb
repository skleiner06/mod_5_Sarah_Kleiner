```python
my_list = [20, 30, 40, 50]

# range(3) ==> [0, 1, 2, 3]
print(f"length of list = {len(my_list)}")

for index in range(len(my_list)):
  print(f"my index is {index} this is my value at that location {my_list[index]}")
```

```python
num_entries = int(input("How many numbers are you entering? "))

total = 0
tries = 0

while tries < num_entries:
  value = int(input("Please enter your number: "))
  total += value
  tries += 1

print(f"Average is {total//(tries + 1)}.")
```

```python
names = []
while len(names) < 3:
  answer = input("Please enter a name: ")
  if answer:
    names.append(answer)
    print(f"Current length of list {len(names)} list {names}.")

print(f"Here is my list of names {names}.")
```

```python
names = []
print("Enter names, one per line. Type 'done' to quit. ")

while True:
  answer = input("Please enter a name: ").strip()

  if answer.lower() == 'done':
    break
  if answer == "":
    print("Please give us a name and try again. ")
    continue

  names.append(answer)
  print(names)

print(f"Here is the entire list {names}. ")
```

```python
password = 'WEDNESDAY'

guess = input("Guess the password: ").strip().upper()

while guess != password:
  print("Nice try! Guess again: ")
  guess = input("Guess the password: ").strip().upper()

print("Great job! Access is granted. ")

```
